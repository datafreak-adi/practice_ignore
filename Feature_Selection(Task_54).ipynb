## Duplicate Method

**Scenario 1:** If the DataFrame contains *columns with identical values but different names*,
 **duplicate_remove** will consider them duplicates, but **duplicated()** won't.

**Scenario 2:** If the DataFrame contains *columns with the same name but different values*,  **duplicate_remove** won't consider them duplicates, but **duplicated()** will.

```python
# Removing duplicate columns

#### Method 1

def duplicate_remove(df):

  duplicate_col={}
  seen_already={}

  for column in df.columns:

    current_column= df[column]

    # convert column to bytes
    try:

      column_hash_code= current_column.values.tobytes()

    except AttributeError:

      column_has_code= current_column.to_string().encode()

    if column_hash_code in seen_already:

      if seen_already[column_hash_code] in duplicate_col:

         duplicate_col[seen_already[column_hash_code]].append(column)

      else:

         duplicate_col[seen_already[column_hash_code]]= [column]


    else:

        seen_already[column_hash_code]= column


  return duplicate_col



```

## Variance Threshold

```python
from sklearn.feature_selection import VarianceThreshold as VT

sel= VT(threshold=0.01)

sel.fit(X1_train)

# to get total columns whose variance greater than threshold (jinko hume rakhna hai)

print(f" Number of colums whose variance threshold greater than  {sum(sel.get_support())}")

# to get name of columns whose variance less than threshold

columns_name= X1_train.columns[sel.get_support()]

values= sel.transform(X1_train)


data_vt= pd.DataFrame(values, columns=columns_name)
```

## Correlation

##Key takes:


**[1]  np.where()**

Expects a string or boolean value. Therefore we use astype(np.bool_)

**[2] np.triu()**

Return upper triangle

**Parameters:**

**m:** array-like
The input array, usually a 2D matrix.

**k:** int, optional
The diagonal above or below which to zero elements.

*k = 0* (default) corresponds to the main diagonal.

[[1 2 3]

 [0 5 6]

 [0 0 9]]


*k > 0* selects diagonals above the main diagonal.


[[0 2 3]

 [0 0 6]

 [0 0 0]]



*k < 0* includes more of the lower triangle by shifting the boundary below the main diagonal.


**NOTE :**   Use **np.tril()** for lower matrix








```python
corr_matrix = data_vt.corr().abs()
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(np.bool_))
to_drop = [column for column in upper_tri.columns if any(upper_tri[column] > 0.8)]
data_corr = data_vt.drop(to_drop, axis=1)

# Numbers Of Columns after Variance Threshold Method

print("Number of Columns after Correlation- ", data_corr.shape[1])
```

## ANOVA

**ANOVA - Approach showed in the session can't be done.** Here every rows of the dataset have some NaN values. ANOVA can't be applied on data having NaN values.

**But**

We can still perform ANOVA analysis by considering each column separately.

ANOVA can be applied to each column individually, comparing the target variable which is Pass/Fail column in our case against the non-missing values in that specific column.

We will drop NaN values of each column before performing the ANOVA



```python
from scipy.stats import f_oneway

column_pvalues=[]

alpha=0.05

data_corr["Pass/Fail"]= data_cp.iloc[:,-1]

for column in data_corr.columns:

  column_data= data_corr[column].dropna()

  anova_results= f_oneway(column_data, data_cp.iloc[:, -1])

  if anova_results.pvalue< alpha:
    # fetching name of columns with p_value less than alpha as for value less than alpha we reject null hypothesis which says(No relation btwn X and Y)
    column_pvalues.append((column, anova_results.pvalue))



column_pvalues.sort(key=lambda x: x[-1])

#The for column, _ in ... syntax is used for unpacking the tuples. column gets assigned the first element (column name) and _ gets assigned the second element (p-value){means throwaway}
# if for _, pvalues ; we would throwaway column numner and get pvalues

selected_column= [column for column, _ in column_pvalues[:100]]

data_anova = data_corr[selected_column+ ["Pass/Fail"]]

print(f" Columns left after annova-test are {data_anova.shape[1]}")

data_anova

```

```python
# check weather we still have null values else we will directly check the accuracy
data_anova.describe()
```

```python
# Since aur data_annova still has null values we treat them first


# Filling NaN Values with random
for column in data_anova.columns:

    min_value = data_anova[column].min()
    max_value = data_anova[column].max()

    # Generate random numbers within the range
    random_values = np.random.uniform(min_value, max_value,  data_anova[column].isnull().sum())

    # Create a Series with the random values
    random_series = pd.Series(random_values, index= data_anova[column][ data_anova[column].isnull()].index)

    # Fill NaN values with the random series
    data_anova[column].fillna(random_series, inplace=True)
```
